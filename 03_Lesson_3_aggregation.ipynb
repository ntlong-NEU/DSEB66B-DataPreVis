# Lesson 3: Data Aggregation and Group Operations

```python
import seaborn as sns
planets = sns.load_dataset('planets')
planets.head()
```

```python
planets['method'].unique().shape
```

```python
adict = {}
for method in planets['method']:
    adict[method] = planets[planets['method']==method]['number'].sum()
```

```python
adict
```

```python
ojb1 = planets.groupby('method')
```

```python
type(ojb1)
```

```python
list(ojb1)[0][1]
```

```python
df03 = planets[['method','number']].groupby('method').sum()
df03
```

```python
df03.shape
```
